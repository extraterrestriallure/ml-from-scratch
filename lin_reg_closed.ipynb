```python
# beta-hat = (x^t*x)^-1 * x^t*y -- closed solution on matrixes
import numpy as np

class LinRegClosed():
    def __init__(self):
        self.coef = None   #coefficients in a matrix
        self.intercept = 0.0 # beta-hat is [b beta1 ... betan]

    def fit(self, x, y):
        x = np.array(x)
        y = np.array(y)

        xb = np.c_[np.ones((x.shape[0], 1)), x]
# calculating beta-hat using linear algebra
# array.T for arrays linalg.matrix_transpose(x) for matrices
# @ preferred for 2d array product
        A = np.linalg.inv(xb.T @ xb) @ xb.T @ y
        self.intercept = A[0]
        self.coef = A[1:]
    
    def predict(self, x):
        x = np.array(x)
        return x @ self.coef + self.intercept
        

import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

```python
data = pd.read_csv('penguins.csv')
data.dropna(inplace=True)
y = pd.DataFrame(data['body_mass_g'])
x = pd.DataFrame(data['flipper_length_mm'])


reg = LinearRegression()
reg.fit(x, y)
y_pred2 = reg.predict(x)

my_reg = LinRegClosed()
my_reg.fit(x,y)
y_pred1 = my_reg.predict(x)

print(f"Are the scores (mine) {r2_score(y,y_pred1)} and (sk) {r2_score(y,y_pred2)} the same?")
```
